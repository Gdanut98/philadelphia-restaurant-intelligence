```python
import pandas as pd
import numpy as np
import nltk
import spacy

# Import the restaurant review dataset
reviews = pd.read_csv('data/restaurant_reviews_az.csv')

# Display the first 5 rows
display(reviews.head())

# Display dataset information
reviews.info()

# Display summary statistics
display(reviews.describe(include='all'))
```

```python
print("Dataset Shape:", reviews.shape)
```

## Select 1 and 5 star reviews

```python
reviews_1_5 = reviews[
    (reviews['stars'] == 1) | (reviews['stars'] == 5)
].copy()

# Display the first 5 rows
display(reviews_1_5.head())

# Check the number of 1-star and 5-star reviews
print(reviews_1_5['stars'].value_counts())
```

## Apply necessary text processing techniques techniques

```python
!python -m spacy download en_core_web_lg
```

```python
nlp = spacy.load("en_core_web_lg")

sentences = []
tokens = []
pos_tags = []
lemmas = []
entities = []
dependencies = []

for doc in nlp.pipe(reviews_1_5['text'], batch_size=50):

    sentences.append([sentence.text for sentence in doc.sents])

    tokens.append([token.text for token in doc])

    pos_tags.append([
        (token.text, token.pos_) for token in doc
    ])

    lemmas.append([
        token.lemma_ for token in doc
    ])

    entities.append([
        (ent.text, ent.label_) for ent in doc.ents
    ])

    dependencies.append([
        (token.text, token.dep_, token.head.text)
        for token in doc
    ])

reviews_1_5['sentences'] = sentences
reviews_1_5['tokens'] = tokens
reviews_1_5['pos_tags'] = pos_tags
reviews_1_5['lemmas'] = lemmas
reviews_1_5['entities'] = entities
reviews_1_5['dependencies'] = dependencies

display(reviews_1_5.head())
```

## Find the top 20 frequently used nouns in 1 star reviews and 5 star reviews

```python
from collections import Counter

# Create lists to store nouns
nouns_1_star = []
nouns_5_star = []

# Loop through each review
for index, row in reviews_1_5.iterrows():

    # Combine POS tags with lemmas
    for pos, lemma in zip(row['pos_tags'], row['lemmas']):

        # Check if the word is a noun
        if pos[1] == 'NOUN':

            # Add noun to appropriate list based on star rating
            if row['stars'] == 1:
                nouns_1_star.append(lemma.lower())

            elif row['stars'] == 5:
                nouns_5_star.append(lemma.lower())

# Find the 20 most frequently used nouns
noun_counter_1_star = Counter(nouns_1_star)
noun_counter_5_star = Counter(nouns_5_star)

top_20_nouns_1_star = noun_counter_1_star.most_common(20)
top_20_nouns_5_star = noun_counter_5_star.most_common(20)

# Display results
print("Top 20 Nouns in 1-Star Reviews:")
print(top_20_nouns_1_star)

print("\nTop 20 Nouns in 5-Star Reviews:")
print(top_20_nouns_5_star)
```

## Word Cloud

```python
# Word Cloud for 1-Star Reviews

import matplotlib.pyplot as plt
from wordcloud import WordCloud

wordcloud = WordCloud()

wordcloud.generate_from_frequencies(noun_counter_1_star)

plt.figure(figsize=(10, 5))

plt.imshow(wordcloud, interpolation="bilinear")

plt.axis("off")

plt.title("1-Star Review Nouns")

plt.show()

# Word Cloud for 5-Star Reviews

wordcloud = WordCloud()

wordcloud.generate_from_frequencies(noun_counter_5_star)

plt.figure(figsize=(10, 5))

plt.imshow(wordcloud, interpolation="bilinear")

plt.axis("off")

plt.title("5-Star Review Nouns")

plt.show()
```

## Find the top 20 frequently used named entities from the selected reviews.

```python
named_entities = []

for index, row in reviews_1_5.iterrows():

    for entity, label in row['entities']:

        named_entities.append((entity.lower(), label))

# Create frequency counter
entity_counter = Counter(named_entities)

# Find top 20
top_20_entities = entity_counter.most_common(20)

# Display results
print("Top 20 Named Entities:")
for entity, count in top_20_entities:
    print(entity, count)
```

```python
# Word Cloud for Named Entities

import matplotlib.pyplot as plt
from wordcloud import WordCloud
from collections import Counter

# Create a counter using only the entity names
entity_word_counter = Counter()

for (entity, label), count in entity_counter.items():
    entity_word_counter[entity] += count

# Create word cloud
wordcloud = WordCloud()

wordcloud.generate_from_frequencies(entity_word_counter)

plt.figure(figsize=(10, 5))

plt.imshow(wordcloud, interpolation="bilinear")

plt.axis("off")

plt.title("Named Entities in 1-Star and 5-Star Reviews")

plt.show()
```
